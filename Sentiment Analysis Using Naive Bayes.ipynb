 Text Cleaning 
1. Sample 10000 rows
2. Remove html tags
3. Remove special characters
4. Converting every thing to lower case
5. Removing Stop Words
6. Stemming

```python
df = df.sample(10000)
```

```python
df.shape
```

```python
# NO missing value lies
df.info()
```

```python
df['sentiment'].replace({'positive':1,'negative':0},inplace=True)
```

```python
df.head()
```

```python
import re
clean = re.compile('<.*?>')
re.sub(clean,'',df.iloc[2].review)
```

```python
# Function to clean html tags
def clean_html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean,'',text)
```

```python
df['review'] = df['review'].apply(clean_html)
```

```python
# Converting everything to lower

def convert_lower(text):
    return text.lower()
```

```python
df['review'] = df['review'].apply(convert_lower)
```

```python
# function to remove special character
def remove_special(text):
    x=''

    for i in text:
        if i.isalnum():
            x=x+i
        else:
            x=x + ' '
    return x
```

```python
df['review'] = df['review'].apply(remove_special)
```

```python
# Remove the step words
# It is natural language toolkit Library
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
df
```

```python
from nltk.corpus import stopwords
sw = set(stopwords.words('english'))

def remove_stopwords(text):
    return [word for word in text.split() if word not in sw]
```

```python
df['review'] = df['review'].apply(remove_stopwords)
```

```python
df
```

```python
# Perform Stemming
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
 y =[]
def stem_words(text):
    for i in text:
        y.append(ps.stem(i))
    z = y[:]
    y.clear()
    return z
```

```python
stem_words(['I','loved','loving','it'])
```

```python
df['review']= df['review'].apply(stem_words)
```

```python
df
```

```python
# Join back
def join_back(list_input):
    return " ".join(list_input)
```

```python
df['review'] = df['review'].apply(join_back)
```
